```python
# Lists - mutable, ordered
features = [0.5, 1.2, -0.3, 0.8, 1.1]
features.append(2.0)
features.sort(reverse=True)
print('Sorted features:', features)

# Tuple - immutable record (sample_id, label, confidence)
record = ('img_001', 'cat', 0.97)
print(f'Sample: {record[0]}, Label: {record[1]}, Conf: {record[2]:.2%}')

# Set algebra for feature selection
train_features = {'age', 'income', 'score', 'clicks', 'region'}
test_features = {'age', 'income', 'score', 'device', 'session'}
common = train_features & test_features
missing = train_features - test_features
print('Common features:', sorted(common))
print('Missing in test:', sorted(missing))

# Frozenset as dictionary key
hashable_key = frozenset({'age', 'income'})
model_cache = {hashable_key: 0.923}
print('Cached accuracy:', model_cache[hashable_key])
```

## Lab 3 - Dictionaries & Nested Structures
**Objective:** Model hierarchical AI dataset schemas using nested dictionaries and lists.

```python
# Nested dataset schema
dataset = {
    'metadata': {'name': 'ImageNet-Mini', 'version': '2.1', 'classes': 10},
    'splits': {
        'train': {'samples': 8000, 'path': '/data/train'},
        'val': {'samples': 1000, 'path': '/data/val'},
        'test': {'samples': 1000, 'path': '/data/test'}
    },
    'class_distribution': {'cat': 1200, 'dog': 1100, 'car': 950}
}

# Nested access
print('Dataset:', dataset['metadata']['name'])
print('Train samples:', dataset['splits']['train']['samples'])

# Dictionary comprehension
normalized = {
    cls: count / sum(dataset['class_distribution'].values())
    for cls, count in dataset['class_distribution'].items()
}
for cls, pct in normalized.items():
    print(f'  {cls}: {pct:.2%}')

# Safe access with .get()
lr = dataset.get('hyperparams', {}).get('learning_rate', 0.001)
print(f'Learning rate: {lr}')
```

## Lab 4 - Lambda, Comprehensions & Functional Patterns
**Objective:** Apply functional programming patterns for ML data pipelines.

```python
import math
from functools import reduce

# Lambda for normalization
min_max_norm = lambda x, mn, mx: (x - mn) / (mx - mn)
data = [10, 25, 5, 40, 15, 30]
min_v, max_v = min(data), max(data)
normalized = [round(min_max_norm(x, min_v, max_v), 4) for x in data]
print('Normalized:', normalized)

# List comprehension with condition (feature filter)
feature_matrix = [[0.1, -0.5, 2.3], [0.8, 1.2, -0.1], [3.1, 0.4, 0.9]]
filtered = [row for row in feature_matrix if max(row) < 3.0]
print('Filtered rows:', filtered)

# Dict comprehension for label encoding
labels = ['cat', 'dog', 'car', 'cat', 'dog']
unique_labels = sorted(set(labels))
encoder = {label: idx for idx, label in enumerate(unique_labels)}
encoded = [encoder[l] for l in labels]
print('Encoder:', encoder)
print('Encoded:', encoded)

# Reduce for cumulative product (learning rate schedule)
lr_decay = [0.9, 0.9, 0.9, 0.9, 0.9]
final_lr = 0.01 * reduce(lambda a, b: a * b, lr_decay)
print(f'LR after 5 epochs: {final_lr:.6f}')
```
