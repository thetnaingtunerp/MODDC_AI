##  Concept / Theory

##  Practical Lab 1 – Lists & List Operations

### Objective
Create, modify, slice, and iterate over lists.

### Explanation
A **list** is an ordered, mutable sequence. In ML, lists hold feature values, class labels, sample indices, and training loss history.

**Key methods:** `append()`, `insert()`, `remove()`, `pop()`, `sort()`, `reverse()`

````python
# Practical Lab 1: Lists
subjects = ['Python', 'Machine Learning', 'Deep Learning', 'Statistics']
print('Initial:', subjects)

# Access
print('First:', subjects[0])
print('Last:', subjects[-1])
print('Slice [1:3]:', subjects[1:3])

# Modify
subjects.append('NLP')           # add to end
subjects.insert(2, 'Math')       # insert at index 2
subjects.remove('Statistics')    # remove by value
print('Modified:', subjects)

# Sort & count
subjects.sort()
print('Sorted:', subjects)
print('Count:', len(subjects))
````

````python
# List operations with scores
scores = [85, 92, 78, 95, 60, 72, 88]
print(f'Scores: {scores}')
print(f'Sum: {sum(scores)}')
print(f'Average: {sum(scores)/len(scores):.2f}')
print(f'Max: {max(scores)} | Min: {min(scores)}')

# Using built-ins
passing = [s for s in scores if s >= 80]
print(f'Passing (>=80): {passing}')
````

### Expected Output
```
Initial: ['Python', 'Machine Learning', 'Deep Learning', 'Statistics']
Sorted: ['Deep Learning', 'Machine Learning', 'Math', 'NLP', 'Python']
Passing (>=80): [85, 92, 95, 88]
```

##  Practical Lab 2 – Tuples & Sets

### Objective
Work with immutable tuples and unique-value sets.

### Explanation
- **Tuple**: ordered, immutable — perfect for image dimensions `(224, 224, 3)` and coordinate pairs.
- **Set**: unordered, unique values — ideal for NLP vocabulary, deduplication, and set math.

````python
# Lab 2: Tuples
student = ('Su Su', 22, 'CS', 91.5)
name, age, dept, gpa = student          # unpacking
print('Record:', student)
print(f'Name={name}, Age={age}, Dept={dept}, GPA={gpa}')

# Image shape (height, width, channels) — common in CV
shape = (224, 224, 3)
h, w, c = shape
print(f'\nImage {w}x{h} with {c} channel(s)')

# Immutability
try:
    student[0] = 'New'
except TypeError as e:
    print(f'Error (expected): {e}')
````

````python
# Lab 2: Sets
cities = ['Yangon','Mandalay','Yangon','Naypyidaw','Mandalay','Bagan']
uniq = set(cities)
print('With duplicates:', cities)
print('Unique set:', uniq)
print('Count:', len(uniq))

# Set operations
python_sk = {'Python','Pandas','NumPy','Matplotlib'}
ml_sk     = {'Pandas','Scikit-learn','NumPy','TensorFlow'}
print('\nUnion:', python_sk | ml_sk)
print('Intersection:', python_sk & ml_sk)
print('Python only:', python_sk - ml_sk)
````

### Expected Output
```
Unique set: {'Bagan', 'Naypyidaw', 'Mandalay', 'Yangon'}
Intersection: {'NumPy', 'Pandas'}
```

##  Practical Lab 3 – Dictionaries

### Objective
Store, access, update, and iterate over key-value pairs.

### Explanation
Dictionaries map keys → values. In ML they hold model hyperparameters, label-to-integer mappings, evaluation metrics, and configuration files.

````python
# Lab 3: Dictionaries
student = {
    'name': 'Mya',
    'age': 21,
    'dept': 'IT',
    'scores': {'Python': 88, 'Math': 92, 'Stats': 85}
}
print('Name:', student['name'])
print('Python score:', student['scores']['Python'])

# Add / update
student['email'] = 'mya@ai.edu'
student['dept']  = 'Computer Science'
print('\nAll fields:')
for k, v in student.items():
    print(f'  {k}: {v}')

# Safe access
phone = student.get('phone', 'Not provided')
print('\nPhone:', phone)
````

````python
# Dictionaries with a list of students
students = [
    {'id':101,'name':'Aung', 'dept':'IT','score':85},
    {'id':102,'name':'Su Su','dept':'IT','score':90},
    {'id':103,'name':'Mya', 'dept':'Biz','score':78},
    {'id':104,'name':'Hla', 'dept':'IT','score':92},
    {'id':105,'name':'Ko Ko','dept':'Biz','score':88},
]
avg = sum(s['score'] for s in students) / len(students)
top = max(students, key=lambda s: s['score'])
print(f'Average: {avg:.1f}')
print(f'Top: {top["name"]} with {top["score"]}')

# Group by dept
dept = {}
for s in students:
    dept.setdefault(s['dept'], []).append(s['score'])
for d, sc in dept.items():
    print(f'{d} avg: {sum(sc)/len(sc):.1f}')
````

### Expected Output
```
Average: 86.6
Top: Hla with 92
IT avg: 89.0
Biz avg: 83.0
```

## Practical Lab 4 – Functions & Lambda

### Objective
Write reusable functions and inline lambda expressions.

### Explanation
- **Functions** (`def`) encapsulate reusable logic — essential for clean ML pipelines.
- **Lambda** creates one-line anonymous functions.
- **Default arguments** make functions flexible.

````python
# Lab 4: Functions
def classify(score):
    """Return performance label."""
    if score >= 90: return 'Excellent'
    elif score >= 80: return 'Good'
    elif score >= 60: return 'Pass'
    else: return 'Fail'

def stats(scores):
    """Return (mean, min, max)."""
    return sum(scores)/len(scores), min(scores), max(scores)

def greet(name, lang='English'):
    words = {'English':'Hello','Myanmar':'Mingalaba'}
    return f"{words.get(lang,'Hi')}, {name}!"

for s in [95, 82, 67, 45]:
    print(f'  {s} => {classify(s)}')

m, lo, hi = stats([85,92,78,95,60])
print(f'\nMean={m:.1f}  Min={lo}  Max={hi}')
print(greet('Aung'))
print(greet('Su Su', lang='Myanmar'))
````

````python
# Lab 4: Lambda
double   = lambda x: x * 2
normalize= lambda x: x / 100

scores = [85,92,78,95,60,72,88]
scaled  = list(map(normalize, scores))
passing = list(filter(lambda s: s >= 80, scores))
print('Scaled:', [f'{s:.2f}' for s in scaled])
print('Passing:', passing)

students = [{'name':'Ko Ko','score':88},{'name':'Mya','score':92},{'name':'Hla','score':75}]
ranked = sorted(students, key=lambda s: s['score'], reverse=True)
print('Ranked:')
for r, s in enumerate(ranked, 1):
    print(f'  {r}. {s["name"]}: {s["score"]}')
````

### Expected Output
```
Passing: [85, 92, 95, 88]
1. Mya: 92
2. Ko Ko: 88
3. Hla: 75
```

##  Practical Lab 5 – List Comprehensions & Nested Structures

### Objective
Write Pythonic comprehensions and handle nested data.

### Explanation
List comprehensions replace verbose loops. Nested structures (lists of dicts) mirror ML datasets.

````python
# Lab 5: Comprehensions
scores = [85,92,78,95,60,72,88]
normed  = [round(s/100, 2) for s in scores]
passing = [s for s in scores if s >= 80]
grades  = ['Pass' if s >= 60 else 'Fail' for s in scores]
print('Normalized:', normed)
print('Passing:', passing)
print('Grades:', grades)

names = ['Aung','Su Su','Mya','Hla','Ko Ko']
map_  = {n: s for n, s in zip(names, scores[:5])}
print('Score map:', map_)
````

````python
# Nested data (typical ML dataset format)
dataset = [
    {'id':1,'features':[1.2,0.8,3.5],'label':0},
    {'id':2,'features':[2.1,1.5,0.9],'label':1},
    {'id':3,'features':[0.5,2.3,1.7],'label':0},
    {'id':4,'features':[3.2,0.1,2.8],'label':1},
]
X = [r['features'] for r in dataset]
y = [r['label']    for r in dataset]
print('Feature matrix X:')
for row in X: print(' ', row)
print('Labels y:', y)
dist = {l: y.count(l) for l in set(y)}
print('Class distribution:', dist)
````

### Expected Output
```
Labels y: [0, 1, 0, 1]
Class distribution: {0: 2, 1: 2}
```

##  Challenge Task – Student Registry System

Build a complete registry using all four structures:
1. 5+ students with ID, name, dept (str), courses (list), scores (dict)
2. `add_student(registry, student)` 
3. `get_top(registry, n=3)` → top-N by average score
4. `dept_summary(registry)` → avg score per department
5. Print a formatted report

````python
# Challenge Task
registry = [
    {'id':101,'name':'Aung', 'dept':'IT', 'courses':['Python','Math'],   'scores':{'Python':85,'Math':88}},
    {'id':102,'name':'Su Su','dept':'IT', 'courses':['Python','Stats'],  'scores':{'Python':92,'Stats':90}},
    {'id':103,'name':'Mya',  'dept':'Biz','courses':['Finance','Math'],  'scores':{'Finance':78,'Math':82}},
    {'id':104,'name':'Hla',  'dept':'IT', 'courses':['Python','ML'],     'scores':{'Python':95,'ML':91}},
    {'id':105,'name':'Ko Ko','dept':'Biz','courses':['Finance','Stats'], 'scores':{'Finance':88,'Stats':85}},
]

def avg_s(s): v=list(s['scores'].values()); return sum(v)/len(v) if v else 0

def add_student(reg, s): reg.append(s); print(f"Added: {s['name']}")

def get_top(reg, n=3): return sorted(reg, key=avg_s, reverse=True)[:n]

def dept_summary(reg):
    d={}
    for s in reg: d.setdefault(s['dept'],[]).append(avg_s(s))
    print('\n--- Dept Summary ---')
    for dept,avgs in d.items(): print(f'  {dept}: {sum(avgs)/len(avgs):.1f}')

add_student(registry, {'id':106,'name':'Thiri','dept':'IT','courses':['ML'],'scores':{'ML':97}})
print('\n--- Top 3 ---')
for i,s in enumerate(get_top(registry),1):
    print(f'  {i}. {s["name"]:8} | avg={avg_s(s):.1f}')
dept_summary(registry)
````

##  Real-World AI/ML Application

| Structure | ML Use |
|---|---|
| **List** | Feature vectors, batch samples, training loss history |
| **Tuple** | Image shapes `(224,224,3)`, train/test splits |
| **Set** | NLP vocabulary, unique class labels |
| **Dict** | Hyperparameters, label encodings, evaluation metrics |

````python
# Real example: label encoding for NLP
raw = [
    {'text':'I love this','label':'positive'},
    {'text':'Terrible!',  'label':'negative'},
    {'text':'Excellent',  'label':'positive'},
    {'text':'Poor',       'label':'negative'},
]
texts    = [r['text'] for r in raw]
labels   = [r['label'] for r in raw]
label_map= {l:i for i,l in enumerate(sorted(set(labels)))}
encoded  = [label_map[l] for l in labels]
print('Label map:', label_map)
print('Encoded y:', encoded)
vocab = sorted(set(' '.join(texts).lower().split()))
print('Vocab size:', len(vocab))
````
